```python
import pandas as pd

# URL of the CSV file (example)
url = "https://raw.githubusercontent.com/vaastav/Fantasy-Premier-League/master/data/2023-24/gws/merged_gw.csv"


# Read the CSV file into a pandas DataFrame
df = pd.read_csv(url)


df.head()
```

```python
sel = ['name', 'position', 'team', 'total_points', 'GW']
df_s = df[sel]

df_s.groupby('team')['total_points'].sum().sort_values(ascending=False)
```

```python
df_s.groupby('position')['total_points'].sum().sort_values(ascending=False)
```

```python
df_s['position'].value_counts()
df_s.groupby('GW')['position']
```
